## 1. FindMax

### Algorithm

```text
ALGORITHM FindMax(A[0..n-1], n)

// Direction: Find the maximum value in the given array.
// Input: Array A of n numbers
// Output: Maximum value of the array elements

    max ← A[0]

    for i ← 1 to n - 1 do
        if A[i] > max
            max ← A[i]

    return max
```

````python
A = [12, 5, 27, 8, 19, 3, 15]

max_value = A[0]

for i in range(1, len(A)):
    if A[i] > max_value:
        max_value = A[i]

print("Array:", A)
print("Maximum value:", max_value)
````

## 2. Average

### Algorithm

```text
ALGORITHM Average(A[0..n-1], n)

// Direction: Calculate the average value of the array elements.

// Input: Array A of n numbers
// Output: Average value of the numbers

    sum ← 0

    for i ← 0 to n - 1 do
        sum ← sum + A[i]

    average ← sum / n

    return average
```

````python
A = [12, 5, 27, 8, 19, 3, 15]

sum_value = 0

for i in range(len(A)):
    sum_value = sum_value + A[i]

average = sum_value / len(A)

print("Array:", A)
print("Average value:", round(average, 2))
````

## 3. Ex. 1.1.4 — IntegerSquareRoot

### Algorithm

```text
ALGORITHM IntegerSquareRoot(n)

// Direction: Find the integer part of the square root of n.

// Input: A positive integer n
// Output: The integer part of √n

    i ← 0

    while (i + 1) × (i + 1) ≤ n do
        i ← i + 1

    return i
```

````python
n = 20
i = 0

while (i + 1) * (i + 1) <= n:
    i = i + 1

print("n =", n)
print("Integer square root:", i)
````

## 4. Ex. 1.2.4 — FindRealRoots

### Algorithm

```text
ALGORITHM FindRealRoots(a, b, c)

// Direction: Find the real roots of the quadratic equation ax² + bx + c = 0.

// Input: Real coefficients a, b, and c
// Output: Real roots of the equation

    if a = 0
        x ← -c / b
        return x

    D ← b × b - 4 × a × c

    if D < 0
        return "No real roots"

    if D = 0
        x ← -b / (2 × a)
        return x

    x1 ← (-b + sqrt(D)) / (2 × a)
    x2 ← (-b - sqrt(D)) / (2 × a)

    return x1, x2
```

````python
import math

a = 1
b = -5
c = 6

if a == 0:
    if b == 0:
        print("Not a valid linear equation")
    else:
        x = -c / b
        print("Root:", x)
else:
    D = b * b - 4 * a * c

    if D < 0:
        print("No real roots")
    elif D == 0:
        x = -b / (2 * a)
        print("Root:", x)
    else:
        x1 = (-b + math.sqrt(D)) / (2 * a)
        x2 = (-b - math.sqrt(D)) / (2 * a)
        print("Roots:", x1, "and", x2)
````

## 5. Ex. 1.2.5 — DecimalToBinary

### Algorithm

```text
ALGORITHM DecimalToBinary(n)

// Direction: Convert a positive decimal integer into its binary representation.

// Input: A positive decimal integer n
// Output: The binary representation of n

    binary ← ""

    while n > 0 do
        remainder ← n mod 2
        binary ← remainder + binary
        n ← n div 2

    return binary
```

````python
n = 25
original_n = n
binary = ""

while n > 0:
    remainder = n % 2
    binary = str(remainder) + binary
    n = n // 2

print("Decimal:", original_n)
print("Binary:", binary)
````

## 7. Ex. 1.2.9 — MinDistance

### Algorithm

```text
ALGORITHM MinDistance(A[0..n-1])

// Direction: Find the minimum distance between two different elements.

// Input: Array A[0..n-1] of numbers
// Output: Minimum distance between two elements

    dmin ← ∞

    for i ← 0 to n - 2 do
        for j ← i + 1 to n - 1 do
            if |A[i] - A[j]| < dmin
                dmin ← |A[i] - A[j]|

    return dmin
```

````python
A = [12, 5, 27, 8, 19, 3, 15]

dmin = float("inf")

for i in range(len(A) - 1):
    for j in range(i + 1, len(A)):
        distance = abs(A[i] - A[j])
        if distance < dmin:
            dmin = distance

print("Array:", A)
print("Minimum distance:", dmin)
````

## 8. Ex. 1.1.6 — Euclid

### Algorithm

```text
ALGORITHM Euclid(a, b)

// Direction: Find the greatest common divisor (GCD) of two positive integers.

// Input: Two positive integers a and b
// Output: Greatest common divisor of a and b

    while b ≠ 0 do
        r ← a mod b
        a ← b
        b ← r

    return a
```

````python
a = 48
b = 18

original_a = a
original_b = b

while b != 0:
    r = a % b
    a = b
    b = r

print("Numbers:", original_a, "and", original_b)
print("GCD:", a)
````

## 9. Ex. 1.1.9 — FindMinMaxDivisions

### Algorithm

```text
ALGORITHM FindMinMaxDivisions

// Direction: Find the minimum and maximum number of divisions
// made by Euclid's algorithm for 1 ≤ m, n ≤ 10.

// Input: Integers m and n from 1 to 10
// Output: Minimum and maximum number of divisions

    minDivisions ← ∞
    maxDivisions ← 0

    for m ← 1 to 10 do
        for n ← 1 to 10 do
            a ← m
            b ← n
            count ← 0

            while b ≠ 0 do
                r ← a mod b
                a ← b
                b ← r
                count ← count + 1

            if count < minDivisions
                minDivisions ← count

            if count > maxDivisions
                maxDivisions ← count

    return minDivisions, maxDivisions
```

````python
min_divisions = float("inf")
max_divisions = 0

for m in range(1, 11):
    for n in range(1, 11):
        a = m
        b = n
        count = 0

        while b != 0:
            r = a % b
            a = b
            b = r
            count += 1

        if count < min_divisions:
            min_divisions = count

        if count > max_divisions:
            max_divisions = count

print("Minimum number of divisions:", min_divisions)
print("Maximum number of divisions:", max_divisions)
````
